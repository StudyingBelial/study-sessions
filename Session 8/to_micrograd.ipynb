```python
class Neuron:
    def __init__(self, nin):
        self.w = [Tensor(1) for _ in range(nin)]
        self.b = Tensor(0.0)

    def __call__(self, xs):
        out = Tensor(0.0)
        out += self.b
        for x, w in zip(xs, self.w):
            out = (x * w) + out
        return out

    def parameters(self):
        params = self.w.copy()
        params.append(self.b)
        return params
```

```python
class Layer:
    def __init__(self, nin, nout):
        self.neurons = [Neuron(nin) for _ in range(nout)]

    def __call__(self, x):
        y = []
        for n in self.neurons:
            y.append(n(x))
        return y

    def parameters(self):
        params = []
        for n in self.neurons:
            params.extend(n.parameters())
        return params
```
